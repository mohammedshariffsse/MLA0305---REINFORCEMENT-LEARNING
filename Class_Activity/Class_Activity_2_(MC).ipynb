Monte Carlo Simulation for Estimation of π

```python
import random
import math
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Number of random points
num_points = 10000

inside_circle = 0
outside_circle = 0

inside_x = []
inside_y = []
outside_x = []
outside_y = []

for i in range(num_points):
    x = random.uniform(-1, 1)
    y = random.uniform(-1, 1)

    if x**2 + y**2 <= 1:
        inside_circle += 1
        inside_x.append(x)
        inside_y.append(y)
    else:
        outside_circle += 1
        outside_x.append(x)
        outside_y.append(y)

estimated_pi = 4 * inside_circle / num_points
actual_pi = math.pi
error = abs(actual_pi - estimated_pi)

print("Estimated Pi :", estimated_pi)
print("Actual Pi    :", actual_pi)
print("Absolute Error :", error)
```

```python
results = pd.DataFrame({
    "Metric": [
        "Total Points",
        "Points Inside Circle",
        "Points Outside Circle",
        "Estimated Pi",
        "Actual Pi",
        "Absolute Error"
    ],
    "Value": [
        num_points,
        inside_circle,
        outside_circle,
        round(estimated_pi,6),
        round(actual_pi,6),
        round(error,6)
    ]
})

results
```

```python
plt.figure(figsize=(6,4))

categories = ["Inside Circle", "Outside Circle"]
values = [inside_circle, outside_circle]

bars = plt.bar(categories, values)

plt.title("Simulation Results Comparison")
plt.xlabel("Category")
plt.ylabel("Number of Points")

for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2,
             height + 50,
             int(height),
             ha='center')

plt.show()
```

```python
plt.figure(figsize=(6,4))

labels = ["Estimated Pi", "Actual Pi"]
values = [estimated_pi, actual_pi]

bars = plt.bar(labels, values)

plt.title("Estimated Pi vs Actual Pi")
plt.ylabel("Pi Value")

for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2,
             height + 0.01,
             f"{height:.5f}",
             ha='center')

plt.show()
```

```python
plt.figure(figsize=(7,7))

plt.scatter(inside_x, inside_y, s=2, label="Inside Circle")
plt.scatter(outside_x, outside_y, s=2, label="Outside Circle")

circle = plt.Circle((0,0),1,color='black',fill=False)
plt.gca().add_patch(circle)

plt.title("Distribution of Random Points")
plt.xlabel("X Coordinate")
plt.ylabel("Y Coordinate")

plt.axis("equal")
plt.legend()

plt.show()
```

```python
steps = []
estimates = []

inside = 0

for i in range(1, num_points + 1):

    x = random.uniform(-1,1)
    y = random.uniform(-1,1)

    if x*x + y*y <= 1:
        inside += 1

    if i % 100 == 0:
        steps.append(i)
        estimates.append(4 * inside / i)

plt.figure(figsize=(8,5))

plt.plot(steps, estimates, linewidth=2, label="Estimated Pi")
plt.axhline(actual_pi, linestyle="--", label="Actual Pi")

plt.title("Convergence of Monte Carlo Estimation")
plt.xlabel("Number of Random Points")
plt.ylabel("Estimated Pi")

plt.legend()
plt.grid(True)

plt.show()
```

Conclusion

The Monte Carlo simulation estimated the value of π by generating random points within a square and determining how many fell inside the inscribed circle. The estimated value closely matched the actual value of π, and the convergence graph showed that the estimate became more accurate as the number of random points increased. This demonstrates the effectiveness of the Monte Carlo method for numerical estimation.
